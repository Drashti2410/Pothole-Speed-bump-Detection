Run Training on Model

```python
from ultralytics import YOLO

# Load the YOLOv11 model
model = YOLO("yolo11l.pt")

# Train the model
model.train(
    data="/content/dataset/data.yaml",  # Update with dataset path
    epochs=100,
    imgsz=640,
    batch=16,
    project="YOLOv11-Training",
    name="pothole-speedbump-detect"
)
```

```python
# Evaluate the model
metrics = model.val()
print(metrics)
```

Testing model on image

```python
from ultralytics import YOLO
import cv2
from matplotlib import pyplot as plt

# Load trained model
model = YOLO("/content/YOLOv11-Training/pothole-speedbump-detect/weights/best.pt")  # Replace with model path best.pt location

# Load the image
img_path = "/content/dataset/train/images/Image00040_jpg.rf.06c3b340e60782c4cc2bc5367e4bb5c4.jpg"  # Replace with test image path
img = cv2.imread(img_path)

# Run inference (stream=True to get result object)
results = model.predict(source=img, conf=0.4, stream=True)

# Plot and show the image with predictions
for r in results:
    annotated = r.plot()  # Annotate the image with boxes/labels
    plt.figure(figsize=(10, 10))
    plt.imshow(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
```

Test model on video

```python
import cv2
from ultralytics import YOLO

# Load the trained model
model = YOLO("/content/best.pt")  # Use correct path to model

# path to video file
cap = cv2.VideoCapture("test_video.mp4")

# Get video properties
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS) or 30  # Default to 30 if not detected

# Define the codec and create VideoWriter object
# saves outputted video to directory with annotated boxes after detetction
out = cv2.VideoWriter('output.mp4',
                      cv2.VideoWriter_fourcc(*'mp4v'),
                      fps,
                      (frame_width, frame_height))

# Loop through frames
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLOv11 prediction
    results = model.predict(source=frame, imgsz=640, conf=0.4, stream=True)

    # Annotate and write to output
    for r in results:
        annotated_frame = r.plot()  # Draw masks and boxes
        out.write(annotated_frame)

# Release everything
cap.release()
out.release()
```
